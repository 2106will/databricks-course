```python
df_customers = spark.read.table("samples.bakehouse.sales_customers")
```

```python
display(df_customers)
```

```python
df_selected = df_customers.select("*")
```

```python
display(df_selected)
```

```python
import pyspark.sql.functions as F
df_filtered = df_customers.filter(F.col("address") == "700 Porter Rapid")
```

```python
display(df_filtered)
```

```python
display(df_selected.filter((F.col("email_address") == "mary39@example.org")& 
                           (F.col("address") == "665 Campbell Streets Suite 966")))
```

```python
display(df_selected.orderBy(F.col("city").desc()).limit(5), F.col("email_address").asc())
```

```python
display(
    df_selected.select(
        F.concat(
            F.col("first_name"),
            F.lit(" "),
            F.col("first_name")            
                ).alias("nombre_completo"),
        F.lit(True).alias("boolean")
    ).limit(5)
)
```

```python
%sql
select country, count(1) as cantidad
from samples.bakehouse.sales_franchises
group by country
order by cantidad desc
```

```python
display(
    df_selected
        .groupBy("country")
        .agg(F.count("*").alias("cantidad"))
        .orderBy(F.col("cantidad").desc())
)
```
